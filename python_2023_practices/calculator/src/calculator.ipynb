```python
"""Examinating the range and sign of the received numbers"""

user_num = input("Enter your number: ")
if user_num.lstrip("-").isdigit():
    user_number = int(user_num)
    if user_number > 0:
        print(user_number, "is a positive number")
    elif user_number < 0:
        print(user_number, "is a negative number")
    else: 
        print(user_num, "is not negative nor positive")
else:
    print("Please enter a number!")    
```

```python
"""Examinating if the received number is odd or even"""

user_num = input("Enter your number: ")
if user_num.lstrip("-").isdigit():
    user_number = int(user_num)
    if user_number %2 == 0:
        print(user_number, "is an even number.")
    else:
        print(user_number, "is an odd number.")
else:
    print("Please enter a number!")
```

```python
"""Score examination"""

st_score = input("Enter your score: ")

if st_score.lstrip("-").isdigit():
    st_s = float(st_score)
    if 0 <= st_s <= 20:

        if st_s >= 18:
            print("You got an 'A'")

        elif 16 <= st_s < 18:
            print("You got a 'B'")

        elif 14 <= st_s < 16:
            print("You got a 'C'")

        elif 12 <= st_s < 14:
            print("You got a 'D'")

        else:
            print("You failed. Repeat the class again!")

        
    else:
        print("Please choose your score between 0 to 20")
```
